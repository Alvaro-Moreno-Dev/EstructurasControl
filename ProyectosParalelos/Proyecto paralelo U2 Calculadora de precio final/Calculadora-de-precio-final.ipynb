# 1. Práctica U2 - 02 - Calculadora de precio final

## 1.1 Objetivo
El objetivo de este ejercicio es desarrollar una aplicación en Python que calcule el **precio final de un producto** a partir de su precio, un descuento y el tipo de IVA aplicable. Para ello, deberás crear una interfaz gráfica utilizando Tkinter (TTK preferentemente) y organizar el código en una clase (no es obligatorio, pero puntúa más). El programa debe permitir al usuario elegir el tipo de IVA, introducir el precio y el descuento, indicar si el precio ya incluye IVA, y mostrar en pantalla el desglose del cálculo.

## 1.2 Requisitos
### 1.2.1. Interfaz gráfica
   - La ventana principal debe tener un título adecuado y un tamaño de 360x400 píxeles.
   - Añade un widget de tipo Sizegrip (TTK) en la esquina inferior derecha con 3 píxeles de margen en todas las direcciones.
   - Añade un marco TTK a la ventana principal con un margen superior de 10 píxeles y un margen lateral de 10 píxeles. Este marco será el de **datos** y contendrá:
        - Un ComboBox TTK de **solo lectura** para seleccionar el tipo de IVA (ver apartado 1.3).
        - Un campo de entrada TTK Entry para el precio en euros.
        - Un Spinbox TTK para el porcentaje de descuento, con valores de 0 a 100 en incrementos de 5 y valor inicial 0.
        - Una etiqueta TTK junto a cada uno de los tres campos anteriores indicando su finalidad. Los campos deben ocupar todo el ancho disponible y guardar márgenes verticales adecuados.
        - Un Checkbutton TTK con el texto "El precio ya incluye IVA", desmarcado por defecto.
   - Añade un segundo marco TTK a la ventana principal con un margen superior de 10 píxeles. Este marco contendrá:
        - Dos botones TTK, "Calcular" y "Limpiar", colocados **uno al lado del otro** y repartiéndose a partes iguales el ancho del marco.
        - Una línea de separación horizontal (TTK) que rellene el ancho del marco, con algo de margen.
        - Una etiqueta TTK donde se mostrará el resultado del cálculo o un mensaje de error. Agrega margen.

### 1.2.2. Funcionalidad
   - Al pulsar "Calcular" se validarán los datos y, si son correctos, se mostrará en la etiqueta el desglose del cálculo en **cinco líneas**: base imponible, descuento aplicado, base con descuento, IVA y total.
   - Todos los importes se mostrarán con **dos decimales** y el símbolo del euro, y deben quedar alineados en columna. Pista: una etiqueta puede usar una fuente de ancho fijo con `font="TkFixedFont"`.
   - El cálculo se realizará así:
        - Si el precio **no** incluye IVA, la base imponible es el precio introducido.
        - Si el precio **sí** incluye IVA, la base imponible es `precio / (1 + IVA)`, donde IVA es el tipo seleccionado expresado en tanto por uno (por ejemplo, 0.21).
        - Importe del descuento = `base × descuento / 100`.
        - Base con descuento = base − importe del descuento.
        - Importe del IVA = base con descuento × IVA.
        - Total = base con descuento + importe del IVA.
   - Al pulsar "Limpiar" se devolverá la aplicación a su estado inicial: ComboBox sin selección, precio vacío, descuento a 0, Checkbutton desmarcado y etiqueta de resultado vacía.
   - Los mensajes de error se mostrarán en color rojo y el resultado correcto en color negro.

### 1.2.3. Clases
   - El código preferentemente estará organizado en una clase llamada `Ventana`. Esta clase debe manejar tanto la creación de la interfaz gráfica como la lógica del cálculo.
   - La clase `Ventana` debe contener métodos para:
        - Inicializar la ventana y los widgets (ComboBox, campos de entrada, botones, etc.).
        - Realizar el cálculo a partir de los datos introducidos y mostrar el resultado.
        - Restablecer la aplicación a su estado inicial.

## 1.3 Tipos de IVA
Los tipos de IVA disponibles en el ComboBox serán los siguientes. Se recomienda almacenarlos en un diccionario que asocie el texto mostrado con el porcentaje.

| Texto del ComboBox     | Porcentaje |
|------------------------|------------|
| General (21%)          | 21         |
| Reducido (10%)         | 10         |
| Superreducido (4%)     | 4          |
| Exento (0%)            | 0          |

## 1.4 Ejemplos de uso
   - IVA General (21%), precio `100`, descuento `10`, sin marcar "El precio ya incluye IVA":
        - Base imponible: 100.00 € · Descuento (10%): 10.00 € · Base con descuento: 90.00 € · IVA (21%): 18.90 € · **TOTAL: 108.90 €**
   - IVA General (21%), precio `121,00`, descuento `0`, con "El precio ya incluye IVA" marcado:
        - Base imponible: 100.00 € · Descuento (0%): 0.00 € · Base con descuento: 100.00 € · IVA (21%): 21.00 € · **TOTAL: 121.00 €**

## 1.5 Validaciones
   - Si no se ha seleccionado un tipo de IVA, el programa debe mostrar un mensaje de error. Emplea, por ejemplo, `mi_etiqueta.config(text="No se ha seleccionado el tipo de IVA", foreground="red")`.
   - Si el precio no es un número, el programa debe mostrar un mensaje de error. Usa `try-except`. Se admitirá tanto el punto como la coma como separador decimal (`12.5` y `12,5` son válidos).
   - Si el precio es negativo, el programa debe mostrar un mensaje de error.
   - Si el descuento no es un número, o no está entre 0 y 100 (ambos incluidos), el programa debe mostrar un mensaje de error. Recuerda que un Spinbox permite escribir texto libre.
   - Ten en cuenta que `float()` acepta cadenas como `"nan"` o `"inf"`: estos valores **no** deben considerarse válidos.

## 1.7 Ampliación opcional
Si has terminado y te sobra tiempo, puedes (sin penalización si no lo haces):
   - Permitir calcular pulsando `Enter` desde el campo de precio.
   - Añadir un cuarto campo "Unidades" y multiplicar el precio por la cantidad antes de aplicar el descuento y el IVA.
   - Mostrar el resultado en verde cuando el total supere los 1000 €.

```python
import tkinter as tk
from tkinter import ttk
import math

class Ventana:
    def __init__(self, root):
        self.root = root
        self.root.title('Calculadora de precios final')
        self.root.geometry('360x400')

        # sizegrip: especie de botón situado en la esquina de la ventana que hace cambiar el tamaño de la ventaña cuando clicamos y arrastramos.
        sizegrip = ttk.Sizegrip(self.root)
        sizegrip.pack(side="bottom", anchor="se", padx=3, pady=3)


        """
        Marco principal de la ventana
        """
        marco_principal = ttk.Frame(self.root) #asociamos el marco a la ventana
        marco_principal.pack(fill="both", expand=True, padx=10, pady=(10, 0))#fill="both", expand=True sirve para que el contenido del marco se adapte al tamaño de la ventana

        """
        Elemento ComboBox del marco principal
        """
        self.tipos_iva = {
            "General (21%)": 21,
            "Reducido (10%)": 10,
            "Superreducido (4%)": 4,
            "Exento (0%)": 0
        }
        ttk.Label(marco_principal, text="Tipo de IVA:").pack(fill="x", pady=(0, 3)) #fill="x" sirve apara que el elemento se ajuste al ancho del marco.
        self.comboBox_tipo_iva = ttk.Combobox(marco_principal, values=list(self.tipos_iva.keys()), state="readonly")
        self.comboBox_tipo_iva.pack(fill="x", pady=(0, 10))

        """
        Campo Entry del marco principal
        """
        ttk.Label(marco_principal, text="Precio en euros:").pack(fill="x", pady=(0, 3))
        self.entrada_precios_euros = ttk.Entry(marco_principal)
        self.entrada_precios_euros.pack(fill="x", pady=(0, 10))

        """
        SpinBox para el porcentaje de descuento
        """
        ttk.Label(marco_principal, text="Porcentaje de descuento:").pack(fill="x", pady=(0, 3))
        self.spinbox_descuento = ttk.Spinbox(marco_principal, from_=0, to=100, increment=5)
        self.spinbox_descuento.set(0)
        self.spinbox_descuento.pack(fill="x", pady=(0, 10))

        """
        CheckButton incluye o no iva
        """
        self.variable_iva = tk.BooleanVar(value=False)
        ttk.Checkbutton(marco_principal, text="El precio ya incluye IVA", variable=self.variable_iva).pack(fill="x", pady=(0, 10))

        """
        Marco secundario (calcular y limpiar )
        """
        marco_secundario = ttk.Frame(self.root)
        marco_secundario.pack(fill="both", expand=True, padx=10, pady=(10, 0))

        """
        Botones y etiqueta marco secundario
        """
        marco_botones = ttk.Frame(marco_secundario)
        marco_botones.pack(fill="x")

        self.boton_calcular = ttk.Button(marco_botones, text="Calcular", command=self.calcular)
        self.boton_calcular.pack(side="left", fill="x", expand=True, padx=(0, 5))

        self.boton_limpiar = ttk.Button(marco_botones, text="Limpiar", command=self.limpiar)
        self.boton_limpiar.pack(side="left", fill="x", expand=True, padx=(5, 0))

        # Separador horizontal
        separador = ttk.Separator(marco_secundario, orient="horizontal")
        separador.pack(fill="x", pady=15)

        # Etiqueta de resultado
        self.etiqueta_resultado = ttk.Label(marco_secundario, text="", font="TkFixedFont")
        self.etiqueta_resultado.pack(fill="x", pady=(0, 10))


    def calcular(self):

        self.etiqueta_resultado.config(foreground="black")

        # VALIDAR IVA
        seleccion_iva = self.comboBox_tipo_iva.get()

        if not seleccion_iva:
            self.etiqueta_resultado.config(text="Error: No se ha seleccionado el tipo de IVA", foreground="red")
            return #Utilizamos return para cortar el flujo de ejecución de la función y que no se ejecute lo de debajo.

        porcentaje_iva = self.tipos_iva[seleccion_iva] / 100

        # VALIDAR PRECIO
        texto_precio = self.entrada_precios_euros.get().replace(",", ".") # permitimos comas y puntos

        try:
            precio = float(texto_precio)

            if precio < 0 or math.isnan(precio) or math.isinf(precio):
                self.etiqueta_resultado.config(text="Error: El precio no es válido o es negativo", foreground="red")
                return

        except ValueError:
            self.etiqueta_resultado.config(text="Error: El precio debe ser un número", foreground="red")
            return

        # VALIDAR DESCUENTO
        texto_descuento = self.spinbox_descuento.get().replace(",", ".")

        try:
            descuento = float(texto_descuento)

            if descuento < 0 or descuento > 100 or math.isnan(descuento) or math.isinf(descuento):
                self.etiqueta_resultado.config(text="Error: El descuento debe estar entre 0 y 100", foreground="red")
                return

        except ValueError:
            self.etiqueta_resultado.config(text="Error: El descuento debe ser un número válido", foreground="red")
            return

        # OPERACIONES
        if self.variable_iva.get():
            base_imponible = precio / (1 + porcentaje_iva)
        else:
            base_imponible = precio

        importe_descuento = base_imponible * (descuento / 100)
        base_con_descuento = base_imponible - importe_descuento
        importe_iva = base_con_descuento * porcentaje_iva
        total = base_con_descuento + importe_iva

        texto_resultado = (
            f"{'Base imponible:':<22} {base_imponible:>8.2f} €\n"
            f"{f'Descuento ({descuento:g}%):':<22} {importe_descuento:>8.2f} €\n"
            f"{'Base con descuento:':<22} {base_con_descuento:>8.2f} €\n"
            f"{f'IVA ({int(porcentaje_iva * 100)}%):':<22} {importe_iva:>8.2f} €\n"
            f"{'TOTAL:':<22} {total:>8.2f} €"
        )

        self.etiqueta_resultado.config(text=texto_resultado, foreground="black")


    def limpiar(self):

        self.comboBox_tipo_iva.set('')
        self.entrada_precios_euros.delete(0, tk.END)
        self.spinbox_descuento.set(0)
        self.variable_iva.set(False)
        self.etiqueta_resultado.config(text="")


if __name__ == "__main__":
    ventana_principal = tk.Tk()
    app = Ventana(ventana_principal)
    ventana_principal.mainloop()
```
